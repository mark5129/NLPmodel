# XLSX to CSV converter

The data for the project is recieved in the format of an excel file containing all the proffesional media articles as well as the regional articles. We want to convert each of these articles to csv files that can be loaded to the various models

**Format in xlsx**

Every row represents a text with various columns representing some information

id | date | title | outlet | link | full text

this has to be converted to an csv file with the same format

```python
# make code that converts an xlsx file to csv. the xlsx file contains two sheets. each sheet represents one area of articles first sheet being "Professional media" and the other being "Regional Media". 
# The xlsx file is named "EnergyIslands_Pro+LocalMediaCoverage_Clean.xlsx" and the csv files should be named "pro_media.csv" and "reg_media.csv" respectively. 
# The csv files should be saved in the same directory as the xlsx file and should have the same rows and columns as the respective sheet.

import pandas as pd

# data/EnergyIslands_Pro+LocalMediaCoverage_Clean.xlsx
# the file is loated in the data folder in a git repository



# read the xlsx file
xlsx = pd.ExcelFile("EnergyIslands_Pro_LocalMediaCoverage_Clean.xlsx")


# read the first sheet
pro_media = pd.read_excel(xlsx, "Professional Media")

# read the second sheet
reg_media = pd.read_excel(xlsx, "Regional Media")

# save the first sheet as a csv file
pro_media.to_csv("pro_media.csv", index=False)

# save the second sheet as a csv file
reg_media.to_csv("reg_media.csv", index=False)

print("Done!")
```
