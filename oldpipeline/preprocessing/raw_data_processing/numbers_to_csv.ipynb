## Convert from xlsx to csv for sci media

```python
import pandas as pd

# Example usage
numbers_file_path = 'EnergyIslands_ScopusAbstracts.xlsx'
csv_file_path = 'EnergyIslands_ScopusAbstracts.csv'
```

```python

# read the xlsx file
xlsx = pd.ExcelFile(numbers_file_path)

# Read the .numbers file
df = pd.read_excel(xlsx)

# Write the dataframe to a .csv file
df.to_csv(csv_file_path, index=False)

print(f"File converted successfully and saved as {csv_file_path}")

```

```python
df = pd.read_csv(csv_file_path)
```

```python
print(df.columns)
```

```python
# Select specific columns to keep
# ['Date', 'Title', 'Outlet', 'Content']
columns_to_keep = ['Year', 'Title','Source title', 'Abstract']
df = df[columns_to_keep]

print(df.columns)
```

```python
df.rename(columns={'Abstract': 'Content'}, inplace=True)
df.rename(columns={'Year': 'Date'}, inplace=True)
df.rename(columns={'Source title': 'Outlet'}, inplace=True)
print(df.columns)
```

```python
df.to_csv('sci_media.csv', index=False)
```
